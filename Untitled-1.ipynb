```python
import os
import torch
import numpy as np
from tqdm import tqdm
from argoverse.map_representation.map_api import ArgoverseMap


# =========================================
# Coordinate Recovery (Option 1)
# =========================================



def to_world_if_possible(pos, data):
    """
    Try to convert local trajectory to Argoverse world frame
    using common preprocessing fields.
    """

    # ---------- origin + theta ----------
    if hasattr(data, "origin") and hasattr(data, "theta"):
        origin = np.array(data.origin)
        theta = float(data.theta)

        R = np.array([
            [np.cos(theta), -np.sin(theta)],
            [np.sin(theta),  np.cos(theta)]
        ])

        return (R.T @ pos[:, :2].T).T + origin

    # ---------- center only ----------
    if hasattr(data, "center"):
        return pos[:, :2] + np.array(data.center)

    # ---------- av_position ----------
    if hasattr(data, "av_position"):
        return pos[:, :2] + np.array(data.av_position)

    # ---------- already world ----------
    return pos[:, :2]


# =========================================
# Maneuver Classifier (STRICT LOGIC)
# =========================================

class ManeuverClassifier:

    def __init__(self):
        self.am = ArgoverseMap()

    # ---------- lane matching ----------
    def _get_best_lane(self, position, city):

        lane_ids = self.am.get_lane_ids_in_xy_bbox(
            position[0],
            position[1],
            city,
            query_search_range_manhattan=5.0
        )

        if not lane_ids:
            return None

        best = None
        best_d = 1e9

        for lid in lane_ids:
            cl = self.am.get_lane_segment_centerline(lid, city)
            d = np.min(np.linalg.norm(cl[:, :2] - position, axis=1))
            if d < best_d:
                best_d = d
                best = lid

        return best


    def lateral_offset(self, lane_id, point, city):
        
        cl = self.am.get_lane_segment_centerline(lane_id, city)[:, :2]

        d = cl - point
        idx = np.argmin(np.linalg.norm(d, axis=1))

        if idx == len(cl) - 1:
            idx -= 1

        p1 = cl[idx]
        p2 = cl[idx + 1]

        v = p2 - p1
        n = np.array([-v[1], v[0]])
        n = n / (np.linalg.norm(n) + 1e-6)

        return np.dot(point - p1, n)


    # ---------- strict classifier ----------
    def classify(self, traj, city, stationary_thresh=2.0):

        # ===== 1. STATIONARY =====
        disp = np.linalg.norm(traj[-1] - traj[0])
        if disp < stationary_thresh:
            return "STATIONARY"

        # ===== 2. START / END LANE =====
        s_lane = self._get_best_lane(traj[0], city)
        e_lane = self._get_best_lane(traj[-1], city)

        if s_lane is None or e_lane is None:
            return "STRAIGHT"

        # ===== 3. TURN CHECK =====
        try:
            turn = self.am.get_lane_turn_direction(e_lane, city)
            if turn == "LEFT":
                return "TURN_LEFT"
            if turn == "RIGHT":
                return "TURN_RIGHT"
        except:
            pass

        # ===== 4. LANE CHANGE CHECK =====
        if s_lane == e_lane:

            try:
                o1 = self.lateral_offset(s_lane, traj[0], city)
                o2 = self.lateral_offset(s_lane, traj[-1], city)
                shift = o2 - o1

                if shift > 1.5:
                    return "LANE_CHANGE_LEFT"
                if shift < -1.5:
                    return "LANE_CHANGE_RIGHT"
            except:
                pass

                queue = next_queue
                depth += 1

        # ===== 5. DEFAULT =====
        return "STRAIGHT"


# =========================================
# Dataset Runner
# =========================================

def main():

    processed_dir = "/mount/studenten/projects/rasoulta/dataset/argoverse/train_processed"

    clf = ManeuverClassifier()

    counts = {
        "STATIONARY": 0,
        "TURN_LEFT": 0,
        "TURN_RIGHT": 0,
        "LANE_CHANGE_LEFT": 0,
        "LANE_CHANGE_RIGHT": 0,
        "STRAIGHT": 0
    }

    files = [f for f in os.listdir(processed_dir) if f.endswith(".pt")]

    for f in tqdm(files):

        try:
            data = torch.load(os.path.join(processed_dir, f))
        except:
            continue

        if not hasattr(data, "positions") or not hasattr(data, "agent_index"):
            continue

        pos = data.positions[data.agent_index].numpy()

        if len(pos) < 2:
            continue

        # -------- world frame recovery --------
        traj = to_world_if_possible(pos, data)

        city = str(data.city) if hasattr(data, "city") else "MIA"

        label = clf.classify(traj, city)

        counts[label] += 1

    print("\nClass distribution:")
    for k, v in counts.items():
        print(k, v)


# =========================================

if __name__ == "__main__":
    main()
```

```python
import os
import numpy as np
from collections import Counter
from tqdm import tqdm

import plotly.graph_objects as go
import torch
from argoverse.map_representation.map_api import ArgoverseMap
from argoverse.data_loading.argoverse_forecasting_loader import ArgoverseForecastingLoader


# ==============================
# MAP SETUP
# ==============================

am = ArgoverseMap()


def lane_is_intersection(am, lane_id, city):
    if hasattr(am, "lane_is_in_intersection"):
        return am.lane_is_in_intersection(lane_id, city)
    if hasattr(am, "lane_segment_is_intersection"):
        return am.lane_segment_is_intersection(lane_id, city)
    return False


# ==============================
# MANEUVER CLASSIFIER
# ==============================

class ManeuverClassifier:

    def __init__(self):
        self.am = am

    def classify_trajectory(self, trajectory, city_name, stationary_thresh=2.0):

        # ---------- Stationary ----------
        disp = np.linalg.norm(trajectory[-1] - trajectory[0])
        if disp < stationary_thresh:
            return "STATIONARY"

        # ---------- Map match ----------
        start_lane = self.get_lane(trajectory[0], city_name)
        end_lane = self.get_lane(trajectory[-1], city_name)

        if start_lane is None or end_lane is None:
            return "STRAIGHT"

        # ---------- Turn ----------
        turn = self.am.get_lane_turn_direction(end_lane, city_name)

        if turn == "LEFT":
            return "TURN_LEFT"
        if turn == "RIGHT":
            return "TURN_RIGHT"

        # ---------- Lane change ----------
        if start_lane != end_lane:

            left = self.get_adj(start_lane, city_name, "left")
            right = self.get_adj(start_lane, city_name, "right")

            succ = self.am.get_lane_segment_successor_ids(start_lane, city_name) or []

            for s in succ:
                if end_lane == self.get_adj(s, city_name, "left"):
                    return "LANE_CHANGE_LEFT"
                if end_lane == self.get_adj(s, city_name, "right"):
                    return "LANE_CHANGE_RIGHT"

            if end_lane == left:
                return "LANE_CHANGE_LEFT"
            if end_lane == right:
                return "LANE_CHANGE_RIGHT"

        return "STRAIGHT"

    def get_adj(self, lane, city, side):
        try:
            adj = self.am.get_lane_segment_adjacent_ids(lane, city)
            if adj:
                return adj[0] if side == "left" else adj[1]
        except:
            pass
        return None

    def get_lane(self, point, city):

        lanes = self.am.get_lane_ids_in_xy_bbox(
            point[0], point[1],
            city,
            query_search_range_manhattan=5
        )

        if not lanes:
            return None

        best = None
        best_dist = 1e9

        for lid in lanes:
            cl = self.am.get_lane_segment_centerline(lid, city)
            d = np.min(np.linalg.norm(cl[:, :2] - point, axis=1))
            if d < best_dist:
                best = lid
                best_dist = d

        return best


classifier = ManeuverClassifier()


# ==============================
# VISUALIZATION
# ==============================

def draw_local_map(fig, point, city):

    lane_ids = am.get_lane_ids_in_xy_bbox(
        point[0], point[1],
        city,
        query_search_range_manhattan=80
    )

    for lid in lane_ids:
        try:
            poly = am.get_lane_segment_polygon(lid, city)
        except:
            continue

        xy = poly[:, :2]

        is_int = lane_is_intersection(am, lid, city)

        fill = 'rgba(255,140,0,0.25)' if is_int else 'rgba(160,160,160,0.25)'

        fig.add_trace(go.Scatter(
            x=xy[:,0],
            y=xy[:,1],
            fill="toself",
            fillcolor=fill,
            line=dict(width=1),
            hoverinfo="skip",
            showlegend=False
        ))


def make_plot(traj, city, label, fname):

    fig = go.Figure()

    draw_local_map(fig, traj[0], city)

    fig.add_trace(go.Scatter(
        x=traj[:,0],
        y=traj[:,1],
        mode="lines+markers",
        name=label
    ))

    fig.update_layout(
        title=f"{label} — {fname}",
        xaxis_scaleanchor="y",
        width=700,
        height=700
    )

    return fig


# ==============================
# DATA HELPERS
# ==============================

def get_agent_traj(seq):
    df = seq.seq_df
    agent_df = df[df["OBJECT_TYPE"] == "AGENT"]
    return agent_df[["X", "Y"]].values


# ==============================
# MAIN PIPELINE
# ==============================


def main():
    # Update this path to your processed .pt directory
    data_path = "/mount/studenten/projects/rasoulta/dataset/argoverse/train_processed"
    
    if not os.path.exists(data_path):
        print(f"Error: Path {data_path} does not exist.")
        return

    files = [f for f in os.listdir(data_path) if f.endswith('.pt')]
    
    counts = Counter()
    saved = 0
    os.makedirs("lane_change_plots", exist_ok=True)

    print(f"Scanning {len(files)} .pt files...")

    for fname in tqdm(files):
        try:
            file_path = os.path.join(data_path, fname)
            # weights_only=False is often needed for custom geometric data objects
            data = torch.load(file_path)

            # Fix for 'string indices must be integers'
            # If data is a list (e.g., [Data(...)])
            if isinstance(data, list):
                data = data[0]

            # 1. Extract and convert tensors to numpy
            agent_idx = data['agent_index']
            
            # Extract positions: shape [N_actors, T_steps, 2]
            # Use .detach().cpu() if you are running on a GPU machine
            local_traj = data['positions'][agent_idx]
            if torch.is_tensor(local_traj):
                local_traj = local_traj.numpy()

            origin = data['origin']
            if torch.is_tensor(origin):
                origin = origin.numpy()
                if origin.ndim > 1:
                    origin = origin.flatten()

            theta = data['rotate_angles'][agent_idx]
            if torch.is_tensor(theta):
                theta = theta.item()

            # 2. Map city index back to string if necessary
            city = data['city']
            if isinstance(city, (int, torch.Tensor)):
                # Argoverse 1 typically uses 0: MIA, 1: PIT
                city_map = {0: "MIA", 1: "PIT"}
                city = city_map.get(int(city), "MIA")

            # 3. Transform Local coordinates back to Global Map coordinates
            # HiVT normalization uses: local = (global - origin) @ Rotation_Matrix
            # Inverse: global = (local @ Rotation_Matrix_Transpose) + origin
            c, s = np.cos(theta), np.sin(theta)
            rotation_mat = np.array([[c, -s], [s, c]])
            
            global_traj = (local_traj @ rotation_mat.T) + origin

            # 4. Classify
            label = classifier.classify_trajectory(global_traj, city)
            counts[label] += 1

            # 5. Visualize lane changes
            if label.startswith("LANE_CHANGE") and saved < 50:
                seq_id = data['seq_id']
                if torch.is_tensor(seq_id):
                    seq_id = seq_id.item()
                
                fig = make_plot(global_traj, city, label, str(seq_id))
                fig.write_html(f"lane_change_plots/{seq_id}.html")
                saved += 1

        except Exception as e:
            # This catches corrupted files or missing keys without stopping the whole script
            print(f"Skipping {fname} due to error: {e}")
            continue

    print("\nClass distribution:")
    if not counts:
        print("No trajectories were classified. Check your coordinate transformation.")
    else:
        for k, v in counts.items():
            print(f"{k}: {v}")


# ==============================
if __name__ == "__main__":
    main()
```

```python
import os
import glob
import numpy as np
import torch
from nuscenes.nuscenes import NuScenes
from nuscenes.map_expansion.map_api import NuScenesMap
from nuscenes.map_expansion import arcline_path_utils
from shapely.geometry import LineString, Point
import plotly.graph_objects as go

# -----------------------
# CONFIGURATION
# -----------------------
NUSCENES_ROOT = '/mount/arbeitsdaten/analysis/rasoulta/nuscenes/nuscenes_meta'
MAP_NAME = 'boston-seaport'
PT_FILES_DIR = '/mount/studenten/projects/rasoulta/dataset/train_processed'
OUTPUT_HTML_DIR = 'nuscenes_map_html'
SEARCH_RADIUS = 50  # meters

os.makedirs(OUTPUT_HTML_DIR, exist_ok=True)

# -----------------------
# LOAD NuScenes + MAP
# -----------------------
nusc = NuScenes(version='v1.0-trainval', dataroot=NUSCENES_ROOT, verbose=False)
nusc_map = NuScenesMap(dataroot=NUSCENES_ROOT, map_name=MAP_NAME)

# -----------------------
# UTILITY FUNCTIONS
# -----------------------
def get_patch_box(x, y, radius):
    """Return square patch coords (xmin, ymin, xmax, ymax)."""
    return (x - radius, y - radius, x + radius, y + radius)

def discretize_lane(lane_token, resolution=1.0):
    """Return list of (x,y) points along lane centerline."""
    arcline = nusc_map.get_arcline_path(lane_token)
    if arcline is None: return []
    poses = arcline_path_utils.discretize_lane(arcline, resolution_meters=resolution)
    return [(p[0], p[1]) for p in poses]

def compute_curvature(poses):
    """Compute average curvature along a lane."""
    pts = np.array(poses)
    if pts.shape[0] < 3: return 0.0
    curvs = []
    for i in range(len(pts) - 2):
        p1, p2, p3 = pts[i], pts[i+1], pts[i+2]
        a = np.linalg.norm(p2 - p1)
        b = np.linalg.norm(p3 - p2)
        c = np.linalg.norm(p3 - p1)
        s = (a+b+c)/2
        area = np.sqrt(max(s*(s-a)*(s-b)*(s-c), 0))
        curv = 0 if area == 0 else (4*area)/(a*b*c)
        curvs.append(curv)
    return float(np.mean(curvs))

def plot_html(ego_pos, ego_theta, elements, html_name):
    """Plot ego, heading, lanes, stop lines, and traffic lights."""
    fig = go.Figure()

    # 1. Lanes
    for lane_token, lane_data in elements['lane'].items():
        xs, ys = zip(*lane_data['poses'])
        fig.add_trace(go.Scatter(x=xs, y=ys, mode='lines', 
                                 line=dict(color='rgba(0, 100, 255, 0.4)', width=2),
                                 name=f'Lane {lane_token[:4]}'))

    # 2. Stop lines (Polygons)
    for sl_coords in elements['stop_line']:
        fig.add_trace(go.Scatter(x=sl_coords[:,0], y=sl_coords[:,1], fill="toself",
                                 line=dict(color='orange', width=2), 
                                 fillcolor='rgba(255, 165, 0, 0.2)', name='Stop Line'))

    # 3. Traffic lights (Lines/Points)
    for tl_coords in elements['traffic_light']:
        fig.add_trace(go.Scatter(x=tl_coords[:,0], y=tl_coords[:,1], mode='markers+lines',
                                 marker=dict(color='green', size=6), name='Traffic Light'))

    # 4. Ego and Heading
    fig.add_trace(go.Scatter(x=[ego_pos[0]], y=[ego_pos[1]],
                             mode='markers', marker=dict(color='red', size=12, symbol='diamond'),
                             name='Ego Position'))
    
    # Heading vector visualization
    dx, dy = np.cos(ego_theta) * 5, np.sin(ego_theta) * 5
    fig.add_annotation(x=ego_pos[0] + dx, y=ego_pos[1] + dy,
                       ax=ego_pos[0], ay=ego_pos[1],
                       xref="x", yref="y", axref="x", ayref="y",
                       showarrow=True, arrowhead=2, arrowsize=1, arrowwidth=3, arrowcolor="red")

    fig.update_layout(title=f'NuScenes Map View - {os.path.basename(html_name)}',
                      xaxis_title='UTM X', yaxis_title='UTM Y',
                      width=900, height=900, xaxis_scaleanchor="y", template="plotly_white")
    fig.write_html(html_name)

# -----------------------
# MAIN LOOP
# -----------------------
pt_files = sorted(glob.glob(os.path.join(PT_FILES_DIR, "*.pt")))[:10]

for idx, pt_file in enumerate(pt_files):
    print(f"\nProcessing: {os.path.basename(pt_file)}")
    data = torch.load(pt_file)
    if isinstance(data, list): data = data[0]

    # Coordinate Extraction
    agent_idx = data['agent_index']
    origin = data['origin'].numpy().flatten()
    theta = data['rotate_angles'][agent_idx].item() if torch.is_tensor(data['rotate_angles']) else data['rotate_angles'][agent_idx]
    local_curr = data['positions'][agent_idx][19].numpy()

    # Transformation to Global UTM
    c, s = np.cos(theta), np.sin(theta)
    rot = np.array([[c, -s], [s, c]])
    global_curr = (local_curr @ rot.T) + origin
    gx, gy = global_curr[0], global_curr[1]

    # Map Query
    patch = get_patch_box(gx, gy, SEARCH_RADIUS)
    records = nusc_map.get_records_in_patch(patch, ['lane', 'stop_line', 'traffic_light'], mode='intersect')

    elements = {'lane': {}, 'stop_line': [], 'traffic_light': []}

    # Process Lanes
    for token in records['lane']:
        poses = discretize_lane(token)
        if poses:
            elements['lane'][token] = {'poses': poses, 'curvature': compute_curvature(poses)}

    # Process Stop Lines (using exterior_node_tokens)
    for token in records['stop_line']:
        record = nusc_map.get('stop_line', token)
        nodes = [nusc_map.get('node', n) for n in record['exterior_node_tokens']]
        pts = np.array([[n['x'], n['y']] for n in nodes])
        if len(pts) > 0:
            pts = np.vstack([pts, pts[0]]) # Close polygon
            elements['stop_line'].append(pts)

    # Process Traffic Lights (using node_tokens)
    for token in records['traffic_light']:
        record = nusc_map.get('traffic_light', token)
        nodes = [nusc_map.get('node', n) for n in record['node_tokens']]
        pts = np.array([[n['x'], n['y']] for n in nodes])
        if len(pts) > 0:
            elements['traffic_light'].append(pts)

    # Export
    html_file = os.path.join(OUTPUT_HTML_DIR, f"view_{idx+1}.html")
    plot_html((gx, gy), theta, elements, html_file)
    print(f"Done. Saved to {html_file}")
```

```python
import os
import glob
import json
import numpy as np
import torch
from nuscenes.nuscenes import NuScenes
from nuscenes.map_expansion.map_api import NuScenesMap
from nuscenes.map_expansion import arcline_path_utils
import plotly.graph_objects as go

# -----------------------
# CONFIGURATION
# -----------------------
NUSCENES_ROOT = '/mount/arbeitsdaten/analysis/rasoulta/nuscenes/nuscenes_meta'
MAP_NAME = 'boston-seaport'
PT_FILES_DIR = '/mount/studenten/projects/rasoulta/dataset/train_processed'

nusc_map = NuScenesMap(dataroot=NUSCENES_ROOT, map_name=MAP_NAME)

def get_ego_relative_data(ego_pos, ego_theta, target_pos):
    """Calculates Euclidean distance and ego-relative angle in degrees."""
    d_vec = target_pos - ego_pos
    dist = np.linalg.norm(d_vec)
    
    # Angle of target in global frame
    global_target_angle = np.arctan2(d_vec[1], d_vec[0])
    
    # Normalize relative to ego heading (0 = straight, + = left)
    rel_angle = (global_target_angle - ego_theta + np.pi) % (2 * np.pi) - np.pi
    return float(dist), float(np.degrees(rel_angle))

# -----------------------
# MAIN LOOP
# -----------------------
pt_files = sorted(glob.glob(os.path.join(PT_FILES_DIR, "*.pt")))[:10]

for idx, pt_file in enumerate(pt_files):
    data = torch.load(pt_file)
    if isinstance(data, list): data = data[0]

    # 1. Coordinate Extraction
    agent_idx = data['agent_index']
    origin = data['origin'].numpy().flatten()
    theta = data['rotate_angles'][agent_idx].item()
    local_curr = data['positions'][agent_idx][19].numpy()
    
    # Global Conversion
    c, s = np.cos(theta), np.sin(theta)
    rot = np.array([[c, -s], [s, c]])
    global_pos = (local_curr @ rot.T) + origin
    
    # 2. Map Querying
    closest_lane_token = nusc_map.get_closest_lane(global_pos[0], global_pos[1], radius=4.0)
    patch = (global_pos[0]-40, global_pos[1]-40, global_pos[0]+40, global_pos[1]+40)
    records = nusc_map.get_records_in_patch(patch, ['lane', 'road_segment', 'stop_line', 'traffic_light'], mode='intersect')

    # 3. Construct JSON Structure
    sample_results = {
        "sample_id": idx,
        "file": os.path.basename(pt_file),
        "ego_lane": None,
        "nearby_lanes": [],
        "intersections": [],
        "traffic_controls": {
            "stop_lines": [],
            "traffic_lights": []
        }
    }

    # Lanes & Curvature
    for l_token in records['lane']:
        path = nusc_map.get_arcline_path(l_token)
        poses = np.array(arcline_path_utils.discretize_lane(path, resolution_meters=1.0))
        mid_pt = poses[len(poses)//2][:2]
        dist, rel_angle = get_ego_relative_data(global_pos, theta, mid_pt)
        
        # Heading Alignment (Relative to ego)
        proj_pose, _ = arcline_path_utils.project_pose_to_lane((global_pos[0], global_pos[1], theta), path)
        align_angle = np.degrees((theta - proj_pose[2] + np.pi) % (2 * np.pi) - np.pi)
        
        # Curvature (Degrees per meter)
        curvature_rad = arcline_path_utils.get_curvature_at_distance_along_lane(len(poses)//2, path)
        curvature_deg = np.degrees(curvature_rad)

        lane_info = {
            "token": l_token,
            "dist_m": round(dist, 2),
            "relative_angle_deg": round(rel_angle, 2),
            "alignment_angle_deg": round(align_angle, 2),
            "curvature_deg_m": round(curvature_deg, 4),
            "is_opposite": abs(align_angle) > 90
        }

        if l_token == closest_lane_token:
            sample_results["ego_lane"] = lane_info
        else:
            sample_results["nearby_lanes"].append(lane_info)

    # Intersections
    for rs_token in records['road_segment']:
        rec = nusc_map.get('road_segment', rs_token)
        if rec['is_intersection']:
            nodes = [nusc_map.get('node', n) for n in rec['exterior_node_tokens']]
            center = np.mean([[n['x'], n['y']] for n in nodes], axis=0)
            dist, rel_angle = get_ego_relative_data(global_pos, theta, center)
            sample_results["intersections"].append({
                "token": rs_token,
                "dist_m": round(dist, 2),
                "relative_angle_deg": round(rel_angle, 2)
            })

    # Traffic Controls
    for layer in ['stop_line', 'traffic_light']:
        json_key = "stop_lines" if layer == "stop_line" else "traffic_lights"
        for token in records[layer]:
            rec = nusc_map.get(layer, token)
            node_key = 'exterior_node_tokens' if layer == 'stop_line' else 'node_tokens'
            pts = np.array([[nusc_map.get('node', n)['x'], nusc_map.get('node', n)['y']] for n in rec[node_key]])
            center = np.mean(pts, axis=0)
            dist, rel_angle = get_ego_relative_data(global_pos, theta, center)
            sample_results["traffic_controls"][json_key].append({
                "token": token,
                "dist_m": round(dist, 2),
                "relative_angle_deg": round(rel_angle, 2)
            })

    # Print the JSON for this sample
    print(json.dumps(sample_results, indent=2))
```

```python
import os
import glob
import json
import numpy as np
import torch
from nuscenes.nuscenes import NuScenes
from nuscenes.map_expansion.map_api import NuScenesMap
from nuscenes.map_expansion import arcline_path_utils

# -----------------------
# CONFIGURATION
# -----------------------
NUSCENES_ROOT = '/mount/arbeitsdaten/analysis/rasoulta/nuscenes/nuscenes_meta'
MAP_NAME = 'boston-seaport'
PT_FILES_DIR = '/mount/studenten/projects/rasoulta/dataset/train_processed'

nusc_map = NuScenesMap(dataroot=NUSCENES_ROOT, map_name=MAP_NAME)

def get_ego_relative_data(ego_pos, ego_theta, target_pos):
    """Calculates Euclidean distance and ego-relative angle in degrees."""
    d_vec = target_pos - ego_pos
    dist = np.linalg.norm(d_vec)
    
    # Global angle of the vector pointing to target
    global_target_angle = np.arctan2(d_vec[1], d_vec[0])
    
    # Normalize relative to ego heading (0 = straight, + = left)
    rel_angle = (global_target_angle - ego_theta + np.pi) % (2 * np.pi) - np.pi
    return float(dist), float(np.degrees(rel_angle))

# -----------------------
# MAIN LOOP
# -----------------------
pt_files = sorted(glob.glob(os.path.join(PT_FILES_DIR, "*.pt")))[:10]

for idx, pt_file in enumerate(pt_files):
    data = torch.load(pt_file)
    if isinstance(data, list): data = data[0]

    # 1. Coordinate Extraction
    agent_idx = data['agent_index']
    origin = data['origin'].numpy().flatten()
    theta = float(data['rotate_angles'][agent_idx].item() if torch.is_tensor(data['rotate_angles']) else data['rotate_angles'][agent_idx])
    local_curr = data['positions'][agent_idx][19].numpy()
    
    # Global Conversion
    c, s = np.cos(theta), np.sin(theta)
    rot = np.array([[c, -s], [s, c]])
    global_pos = (local_curr @ rot.T) + origin
    
    # 2. Identify "Home" Lane
    closest_lane_token = nusc_map.get_closest_lane(global_pos[0], global_pos[1], radius=4.0)
    
    patch = (global_pos[0]-50, global_pos[1]-50, global_pos[0]+50, global_pos[1]+50)
    records = nusc_map.get_records_in_patch(patch, ['lane', 'road_segment', 'stop_line', 'traffic_light'], mode='intersect')

    sample_results = {
        "sample_id": int(idx),
        "file": os.path.basename(pt_file),
        "ego_lane": None,
        "nearby_lanes": [],
        "intersections": [],
        "traffic_controls": {"stop_lines": [], "traffic_lights": []}
    }

    # 1. Process Lanes (We keep nearby lanes even if slightly behind for context)
    for l_token in records['lane']:
        path = nusc_map.get_arcline_path(l_token)
        proj_pose, _ = arcline_path_utils.project_pose_to_lane((global_pos[0], global_pos[1], theta), path)
        closest_pt = np.array([proj_pose[0], proj_pose[1]])
        dist_to_lane, pos_angle = get_ego_relative_data(global_pos, theta, closest_pt)
        lane_yaw = proj_pose[2]
        heading_error = np.degrees((theta - lane_yaw + np.pi) % (2 * np.pi) - np.pi)
        
        lane_info = {
            "token": str(l_token),
            "dist_to_closest_pt_m": round(float(dist_to_lane), 2),
            "look_at_lane_angle_deg": round(float(pos_angle), 2),
            "heading_alignment_error_deg": round(float(heading_error), 2),
            "is_oncoming_traffic": bool(abs(heading_error) > 90)
        }
        if l_token == closest_lane_token: sample_results["ego_lane"] = lane_info
        else: sample_results["nearby_lanes"].append(lane_info)

    # 2. Process Intersections & Controls (STRICT FRONT-ONLY FILTER)
    for layer in ['road_segment', 'stop_line', 'traffic_light']:
        for token in records[layer]:
            rec = nusc_map.get(layer, token)
            if layer == 'road_segment' and not rec['is_intersection']: continue
            
            # Find center point of the object
            if layer == 'stop_line': nodes = rec['exterior_node_tokens']
            elif layer == 'traffic_light': nodes = rec['node_tokens']
            else: nodes = rec['exterior_node_tokens']
            
            pts = np.array([[nusc_map.get('node', n)['x'], nusc_map.get('node', n)['y']] for n in nodes])
            center = np.mean(pts, axis=0)
            
            dist, rel_angle = get_ego_relative_data(global_pos, theta, center)
            
            # FILTER: Ignore if angle is > 90 or < -90 (i.e., behind the ego)
            if abs(rel_angle) > 90:
                continue
            
            data_obj = {
                "token": str(token), 
                "dist_m": round(dist, 2), 
                "relative_angle_deg": round(rel_angle, 2)
            }
            
            if layer == 'road_segment': sample_results["intersections"].append(data_obj)
            elif layer == "stop_line": sample_results["traffic_controls"]["stop_lines"].append(data_obj)
            else: sample_results["traffic_controls"]["traffic_lights"].append(data_obj)

    print(json.dumps(sample_results, indent=2))
```
